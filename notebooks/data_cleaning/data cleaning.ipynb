```python
import ast
def eval(x):
    if isinstance(x,str):
        x=x.strip()
        if x.startswith("[") and x.endswith("]"):
            return ast.literal_eval(x)
        return[x]
    return x
```

```python
tmdb_df["genre_labels"]=tmdb_df["genre_labels"].apply(eval)
```

```python
import ast
movielens_df["genre_label"]=movielens_df["genre_label"].apply(
    lambda x:ast.literal_eval(x) if isinstance(x,str) else x)
```
