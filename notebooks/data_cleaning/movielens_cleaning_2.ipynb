```python
import re 
def split_title(title):
    match=re.search(r"\((\d{4})\)",title)
    if match:
        year=int(match.group(1))
    else:
        year=None

    clean_title=re.sub(r"\(\d{4}\)","",title).strip().lower()
    return clean_title,year
                    
```

```python
df[["clean_title","year"]]=df["title"].apply(lambda x:pd.Series(split_title(x)))
```
