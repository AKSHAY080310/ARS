```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(stop_words="english")
tfidf_matrix=tfidf.fit_transform(df["combined_features"])
```

```python
row=tfidf_matrix[0]
indices=row.nonzero()[1]
score=row.data
word_scores=list(zip(indices,score))
word_scores=sorted(word_scores,key=lambda x:x[1],reverse=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(tfidf_matrix)
```

```python
print(similarity[0])
```

```python
def recommend_movie(movie):
    index=df[df["title"]==movie].index[0]
    movie_lang=df.iloc[index]["language"]
    distance=list(enumerate(similarity[index]))
    distances=[(i,score) for i,score in distance if df.iloc[i]["language"]==movie_lang]
    movie_list=sorted(distances,key=lambda x:x[1],reverse=True)[:6]
    for i in movie_list:
        print(df.iloc[i[0]][["title","language"]])
```

```python
recommend_movie("gladiator")
```
