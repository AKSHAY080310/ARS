```python
import pandas as pd
df=pd.read_pickle(r"C:\Users\aksha\OneDrive\Desktop\ARS(advanced recommendation system)\data\processed\movielens_for_collaborative.pkl")
```

```python
user_item_matrix=df.pivot_table(index="user_id",columns="clean_title",values="rating")
```

```python
print(user_item_matrix)
```

```python
user_item_matrix=user_item_matrix.fillna(0)
```

```python
print(user_item_matrix)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
user_similarity=cosine_similarity(user_item_matrix)
```

```python
print(user_similarity)
```

```python
user_similarity_df=pd.DataFrame(user_similarity,index=user_item_matrix.index,columns=user_item_matrix.index)
```

```python
user_similarity_df.head()
```

```python
def recommend_movies(user_id, top_n=5):
    similar_users = user_similarity_df[user_id].sort_values(ascending=False)[1:6]
    print(similar_users)
    similar_users_ratings = user_item_matrix.loc[similar_users.index]
    print(similar_users_ratings)
    recommendation_scores = similar_users_ratings.mean().sort_values(ascending=False)
    print(recommendation_scores)
    watched = user_item_matrix.loc[user_id]
    print(watched)
    recommendation_scores = recommendation_scores[watched == 0]
    print(recommendation_scores)
    return recommendation_scores.head(top_n)
```

```python
print(recommend_movies(1))
```

```python
item_similarity = cosine_similarity(user_item_matrix.T)

item_similarity_df = pd.DataFrame(
    item_similarity,
    index=user_item_matrix.columns,
    columns=user_item_matrix.columns
)
```

```python
def recommend_similar(movie_name, top_n=5):
    return item_similarity_df[movie_name] \
        .sort_values(ascending=False)[1:top_n+1]
```

```python
print(recommend_similar("clear and present danger"))
```
