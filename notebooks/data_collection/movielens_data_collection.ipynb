```python
import pandas as pd

genre_df=pd.read_csv(
    "C:\\Users\\aksha\\OneDrive\\Desktop\\movies\\archive (3)\\ml-100k\\u.genre",
    header=None,
    sep="|",
    names=["genres","genre_id"])
genre_df.head(19)
    
```

```python
cols=["item_id","title","release_date","video_Release_date","imdb_url"]+genre_df["genres"].tolist()
movies_df=pd.read_csv(
    "C:\\Users\\aksha\\OneDrive\\Desktop\\movies\\archive (3)\\ml-100k\\u.item",
    sep="|",
    names=cols,
    header=None,
    encoding="latin-1"
)
movies_df.head()
```

```python
genre_cols=genre_df["genres"].tolist()
```

```python
def extract_genre(row):
    return ",".join([g for g in genre_cols if row[g]==1])
```

```python
movies_df["genre_label"]=movies_df.apply(extract_genre,axis=1)
```

```python
movies_df.head()
```

```python
movies_df.drop(["video_Release_date","imdb_url"], axis=1)
```

```python
movies_df.head()
```

```python
movies_df=movies_df.drop(["video_Release_date","imdb_url"],axis=1)
```

```python
movies_df=movies_df.drop(columns=genre_cols,axis=1)
```

```python
movies_df.head()
```

```python
movies_df["genre_label"] = movies_df["genre_label"].str.replace(",", "|")
```

```python
interactions=pd.read_csv(
    "C:\\Users\\aksha\\OneDrive\\Desktop\\movies\\archive (3)\\ml-100k\\u.data",
    sep="\t",
    header=None,
    names=["user_id","item_id","rating","timestamp"]
)
interactions.head()
```

```python
users=pd.read_csv(
    "C:\\Users\\aksha\\OneDrive\\Desktop\\movies\\archive (3)\\ml-100k\\u.user",
    sep="|",
    names=["user_id","age","gender","occupatioin","zipcode"]
)
users.head()
```

```python
merged_df=interactions.merge(users,on="user_id",how="left")
```

```python
merged_df.head()
```

```python
final_df=merged_df.merge(movies_df,on="item_id",how="left")
```

```python
final_df.head()
```

```python
final_df.drop(["timestamp","zipcode"],axis=1,inplace=True)
```
