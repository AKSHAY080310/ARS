```python
GENRE_NORMALIZATION_MAP = {
    "Sci-Fi": "Science Fiction",
    "Musical": "Music",
    "Children's": "Family"
}
```

```python
def normalize_genre(genres):
    if not isinstance(genres,list):
        return "unknown"
    if len(genres)==0:
        return "unknown"
    normalized={
        GENRE_NORMALIZATION_MAP.get(g,g)
        for g in genres
    }
    return list(normalized)
```

```python
type(tmdb_df["genre_labels"].iloc[0])
```

```python
import ast
tmdb_df["genre_labels"]=tmdb_df["genre_labels"].apply(
    lambda x:ast.literal_eval(x) if isinstance(x,str) else x)
```

```python
tmdb_df["genre_labels"]=tmdb_df["genre_labels"].apply(normalize_genre)
```

```python
def split_movielens_genres(genre_str):
    if genre_str == "(no genres listed)" or not genre_str:
        return ["unknown"]
    return genre_str.split("|")
```

```python
movielens_df["genre_label"] = movielens_df["genre_label"].apply(split_movielens_genres)
```
